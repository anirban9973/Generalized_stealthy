```python
import h5py
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

mpl.rcParams['pdf.fonttype'] = 42
```

```python
h5path = 'configs.h5'

with h5py.File(h5path, 'r') as f:
    basis = f.attrs['basis'][:]   # (dim, dim)
    N     = int(f.attrs['N'])
    pos   = f['config_0'][:]      # (N, dim)

a1 = basis[0]
a2 = basis[1]
```

```python
# Box corners: O -> a1 -> a1+a2 -> a2 -> O
box = np.array([[0,0], a1, a1+a2, a2, [0,0]])

fig, ax = plt.subplots(figsize=(5, 5))

ax.plot(box[:, 0], box[:, 1], 'k-', linewidth=0.8)
ax.scatter(pos[:, 0], pos[:, 1], s=2, c='steelblue', linewidths=0)

ax.set_aspect('equal')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title(f'config_0   (N={N})')

plt.tight_layout()
plt.savefig('config_0.pdf')
plt.show()
```
